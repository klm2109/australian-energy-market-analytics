# Data Cleaning and Preparation

## Australian Energy Market Analytics

This notebook prepares electricity price and demand data from the Australian
Energy Market Operator (AEMO) for exploratory and statistical analysis.

### Objectives

- Load and inspect the raw AEMO price and demand data
- Assess data quality and identify missing or duplicate observations
- Convert timestamps into an appropriate datetime format
- Validate electricity demand and spot price values
- Create time-based features for subsequent analysis
- Export a cleaned dataset for exploratory analysis

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Load the Dataset

The dataset contains five-minute electricity demand and Regional Reference
Price (RRP) observations from the Australian Energy Market Operator (AEMO).

```python
file_path = "../data/raw/PRICE_AND_DEMAND_202608_NSW1.csv"

df = pd.read_csv(file_path)

df.head()
```

## 3. Initial Data Inspection
### 3.1. Dataset Dimensions

```python
df.shape

print(f"Number of observations: {df.shape[0]:,}")
print(f"Number of variables: {df.shape[1]}")
```

The raw dataset contains **7,776** observations across **5** variables.
Each observation represents a five-minute electricity market interval.

### 3.2. Inspect column types

```python
df.info()
```

`SETTLEMENTDATE` is currently stored as an object rather than a datetime
variable. This will be converted before performing time-series analysis.

```python
df["SETTLEMENTDATE"] = pd.to_datetime(df["SETTLEMENTDATE"])
df.info()
```

### Variable Description

| Variable | Description |
|---|---|
| `REGION` | National Electricity Market region identifier |
| `SETTLEMENTDATE` | Timestamp for the five-minute market interval |
| `TOTALDEMAND` | Total regional electricity demand in MW |
| `RRP` | Regional Reference Price in $/MWh |
| `PERIODTYPE` | Type of market interval |

## 4. Data Quality Assessment

### 4.1 Missing Values

```python
missing_values = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": df.isnull().mean() * 100
})

missing_values
```

No missing values were identified in the raw dataset, so no imputation or
row removal is required at this stage.

### 4.2 Duplicate Observations

```python
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

## 5. Export Cleaned Dataset

The cleaned dataset is exported for use in subsequent exploratory
and statistical analyses.

```python
output_path = "../data/processed/clean_price_demand.csv"

df.to_csv(output_path, index=False)

print(f"Cleaned dataset saved to: {output_path}")
```
